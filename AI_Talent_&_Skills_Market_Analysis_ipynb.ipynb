# **Stage - 2 Data Cleaning and Pre-processing**

1. Checked and handled missing values in the dataset.
2. Identified and removed duplicate records.
3. Renamed columns to improve readability.
4. Removed unnecessary columns that were not required for analysis.
5. Checked and converted data types where necessary.
6. Identified salary outliers and retained valid high salary values.
7. Analyzed skewness and applied log transformation to Demand Growth YoY %.
8. Created new features such as Posting Period, Quarter, Salary Range USD, and Month Salary USD.

These steps helped prepare the dataset for Exploratory Data Analysis (EDA) and business insights.


```python
# Check null Values
df.isnull().sum()
```

```python
df.isnull().sum()[df.isnull().sum()>0]
```

```python
# Check duplicated values
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
df.shape
```

```python
# Remove unnecessary columns
df.drop(columns=["job_id"], inplace=True)
```

Statement:

Removed job_id because it is only a unique identifier and is not useful for the analysis.

```python
df[["experience_level","is_senior","remote_work","is_remote_friendly"]].head()
```

```python
df.drop(columns=["is_senior","is_remote_friendly"],inplace=True)
df.columns
```

Cleaning reason:
Removed redundant columns is_senior and is_remote_friendly because the same information is already represented by experience_level and remote_work.

```python
# Rename columns name
df.rename(columns={"company_size":"Company Employee Range"}, inplace=True)
df.rename(columns={"remote_work":"Work Type"}, inplace=True)
df[["Company Employee Range","Work Type"]]
```

```python
df.rename(columns={"salary_tier": "Salary Category"}, inplace=True)
df['Salary Category'] = df['Salary Category'].str.split('(').str[0].str.strip()
df['Salary Category']
```

```python
# Columns name Standardization
df.columns = df.columns.str.title().str.replace("_"," ",regex=False)
df.columns = (
    df.columns
    .str.replace("Usd", "USD", regex=False)
    .str.replace("Ai", "AI", regex=False)
    .str.replace("Llm", "LLM", regex=False)
    .str.replace("Yoy", "YoY", regex=False)
    .str.replace("Pct", "%", regex=False)
    .str.replace("Benefits Score 10", "Benefits Score", regex=False))
df.columns
```

```python
# columns name unwanted Space remove
df.columns = df.columns.str.strip()
```

```python
# Required skills row_wise checking the duplicate

df['Required Skills'].apply(lambda x: pd.Series(x.split('|')).duplicated().any()).sum()
```

```python
# Duplicated skill remove

df['Required Skills'] = df['Required Skills'].apply(lambda x: '|'.join(dict.fromkeys(x.split('|'))))
df['Required Skills'].apply(lambda x: pd.Series(x.split('|')).duplicated().any()).sum()
```

```python
# Unwanted space remove checking
df["Required Skills"]=df["Required Skills"].str.strip()
df["Required Skills"].unique()
```

```python
# convert the data types
df.dtypes
```

```python
# Check the Outlier annual_salary
df["Annual Salary USD"].describe()
```

```python
#IQR - Inter Quartile Range
Q1 = df["Annual Salary USD"].quantile(0.25)
Q3 = df["Annual Salary USD"].quantile(0.75)
IQR = Q3 - Q1
Lower = Q1 - 1.5 * IQR
Upper = Q3 + 1.5 * IQR
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower:", Lower)
print("Upper:", Upper)
```

```python
Outliers = df[(df['Annual Salary USD'] < Lower)|(df['Annual Salary USD'] > Upper)]
print("Num of outliers:",len(Outliers))
```

```python
Outliers[['Annual Salary USD']]
```

Outlier Handling:

IQR method was used and 8 outliers were found. These were valid higher salary values, so they were kept without removing or changing them.


```python
print("Numerical columns Checking Skewness:")
print(df.select_dtypes(include=['float64', 'int64']).skew())
```

Most columns → Skewness < 1 → No transformation needed

Demand Growth YoY % = 1.042 → Skewness > 1 → Check the distribution

Is LLM Role = 0/1 → Binary column → No transformation

Posting Month = 1–12 → Discrete month value → No transformation

```python
# Distribution of the Demand Growth
import matplotlib.pyplot as plt
df["Demand Growth YoY %"].hist()
plt.xlabel("Demand Growth YoY %")
plt.ylabel("Frequency")
plt.show()
```

```python
df['Demand Growth YoY %'].min()
```

```python
#apply transformation
import numpy as np
df["Demand Growth YoY log"]=np.log1p(df["Demand Growth YoY %"])
print("Skewness:",df["Demand Growth YoY log"].skew())
```

Statement:

Demand Growth YoY (%) was positively skewed with a skewness of 1.042. Log transformation was applied to reduce the skewness. After transformation, the skewness reduced to 0.05, indicating a more balanced distribution.

# Feature Transformation:

```python
# Feature Transformation(Date parts)
df["Posting Period"]=pd.to_datetime(df["Posting Year"].astype(str)+"-"+df["Posting Month"].astype(str)+"-01").dt.strftime("%b %Y")
```

```python
# Salary Range
df["Salary Range USD"] = df["Salary Max USD"] - df["Salary Min USD"]
```

```python
# Add columns Annual Salary from Month Salary
df["Monthly Salary USD"] = (df["Annual Salary USD"] / 12).round(2)
```

```python
df["Quarter"] = "Q" + pd.to_datetime(df["Posting Period"]).dt.quarter.astype(str)
```

```python
# Multivariate Analysis → Job Category + Required Skills + Experience Level - groupby analysis
# Which AI job categories have the highest demand for specific skills across different experience levels?

Skill_data = df.assign(Required_Skills=df["Required Skills"].str.split("|")).explode("Required_Skills")
Skill_data["Required_Skills"] = Skill_data["Required_Skills"].str.strip()
group_analysis = (Skill_data.groupby(["Job Category", "Required_Skills","Experience Level"]).size().reset_index(name="Job Count"))
top_10= (group_analysis.sort_values("Job Count", ascending=False).head(10))
top_10
```

```python
#Top Required Skills by Experience Level in AI Engineering Jobs
plt.figure(figsize=(8,5))
sns.barplot(data=top_10,x="Experience Level",y="Job Count",hue="Required_Skills",edgecolor="black",palette="Set2")
plt.title("Top Required Skills by Experience Level and Industry",fontweight="bold")
plt.title("Top Required Skills by Experience Level",fontweight="bold")
plt.xlabel("Experience Level")
plt.ylabel("Job Count")
plt.show()
```

### Interpretation

**Explain the Chart:**

This analysis shows the top combinations of **Job Category, Required Skill, and Experience Level** based on the number of job postings.

**What is it saying?**

```python
Advance_data = (df.groupby(["Industry", "Company Employee Range", "Experience Level"])["Demand Score"].mean().reset_index())

sns.set_theme(style="whitegrid", font_scale=1.1)

r = sns.catplot(data=Advance_data,x="Industry",y="Demand Score",hue="Experience Level",col="Company Employee Range",
    col_wrap=3,kind="bar",height=5.5,aspect=1.4,palette="Set1",sharex=False)

r.set_axis_labels("Industry", "Average Demand Score")
r.set_titles("{col_name}", size=13, weight="bold")

for ax in r.axes.flat:
    ax.tick_params(axis="x", labelbottom=True, rotation=90, labelsize=10)
    ax.tick_params(axis="y", labelsize=10)
    ax.grid(axis="y", linestyle="--", alpha=0.3)

r.figure.subplots_adjust(top=0.90, hspace=0.6, wspace=0.25)

r.figure.suptitle("Industry and Experience-wise Demand Score by Company Size",fontsize=16,fontweight="bold")

plt.show()

```
